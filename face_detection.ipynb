```python
import cv2
```

```python
face_cascade=cv2.CascadeClassifier(r"C:\Users\LoQ\Desktop\AI\(AI) Deep learning\CV\haarcascade_frontalface_default.xml")
```

```python
eye_cascade=cv2.CascadeClassifier(r"C:\Users\LoQ\Desktop\AI\(AI) Deep learning\CV\haarcascade_eye.xml")
```

```python
import cv2

video = cv2.VideoCapture(0)

print("Camera opened:", video.isOpened())

check, frame = video.read()

print("Frame received:", check)

if check:
    cv2.imshow("Camera Test", frame)
    cv2.waitKey(0)

video.release()
cv2.destroyAllWindows()
```

```python
def detect(gray,frame):
    faces = face_cascade.detectMultiScale(gray,1.3,5)
    for (x,y,w,h) in faces:
        cv2.rectangle(frame,(x,y),(x+w,y+h),(255,0,0),2)
        roi_gray = gray[y:y+h, x:x+w]
        roi_color = frame[y:y+h, x:x+w]
        eyes = eye_cascade.detectMultiScale(roi_gray,1.1,3)
        for(ex,ey,ew,eh) in eyes:
            cv2.rectangle(roi_color,(ex,ey),(ex+ew,ey+eh),(0,255,0),2)
    return frame
```

```python
video = cv2.VideoCapture(0)
while True:
    check , frame =video.read()
    gray = cv2.cvtColor(frame , cv2.COLOR_BGR2GRAY)
    canvas = detect(gray,frame)
    cv2.imshow('Video', canvas)
    
    if cv2.waitKey(1) == ord('q'):
        break
video.release()
cv2.destroyAllWindows()
```
